# Exercício 18 — Número Primo

### Enunciado

Crie um programa que verifique se um número fornecido é primo.

### Explicação

Um número primo possui apenas dois divisores: 1 e ele mesmo.

Para verificar isso, testo se existe algum número entre 2 e o número anterior que consiga dividi-lo sem deixar resto.

```python
# Solicita um número inteiro.
numero = int(input("Digite um número: "))

# Assume inicialmente que o número é primo.
primo = True

# Números menores que 2 não são primos.
if numero < 2:
    primo = False
else:
    # Verifica possíveis divisores.
    for i in range(2, numero):
        if numero % i == 0:
            primo = False
            break

# Exibe o resultado.
if primo:
    print("O número é primo.")
else:
    print("O número não é primo.")
```

# Exercício 20 — Matriz Transposta

### Enunciado

Dada uma matriz (lista de listas), crie sua matriz transposta.

### Explicação

A matriz transposta troca as linhas pelas colunas.

Neste exercício, percorro as posições da matriz e construo uma nova lista contendo os elementos organizados dessa forma.

```python
# Cria uma matriz com 2 linhas e 3 colunas.
matriz = [
    [1, 2, 3],
    [4, 5, 6]
]

# Cria a matriz transposta.
transposta = []

for coluna in range(len(matriz[0])):
    nova_linha = []

    for linha in range(len(matriz)):
        nova_linha.append(matriz[linha][coluna])

    transposta.append(nova_linha)

# Exibe a matriz original.
print("Matriz original:")
for linha in matriz:
    print(linha)

# Exibe a matriz transposta.
print("\nMatriz transposta:")
for linha in transposta:
    print(linha)
```

# Exercício 21 — Busca Binária

### Enunciado

Implemente a busca binária para encontrar um elemento em uma lista ordenada.

### Explicação

A busca binária funciona somente com uma lista ordenada. Ela verifica o elemento que está no meio da lista e elimina metade das possibilidades a cada comparação.

O processo continua até encontrar o valor ou até não existirem mais posições para pesquisar.

```python
# Cria uma lista já ordenada.
numeros = [2, 5, 8, 12, 16, 20, 25, 30]

# Solicita o valor que será procurado.
valor = int(input("Digite o número que deseja procurar: "))

# Define os limites da busca.
inicio = 0
fim = len(numeros) - 1
encontrado = False

# Realiza a busca binária.
while inicio <= fim:
    meio = (inicio + fim) // 2

    if numeros[meio] == valor:
        encontrado = True
        break
    elif numeros[meio] < valor:
        inicio = meio + 1
    else:
        fim = meio - 1

# Exibe o resultado.
if encontrado:
    print("Número encontrado na posição:", meio)
else:
    print("Número não encontrado.")
```

# Exercício 22 — Criptografia de Cifra de César

### Enunciado

Crie uma função para criptografar uma mensagem usando a cifra de César.

### Explicação

A cifra de César substitui cada letra por outra letra algumas posições à frente no alfabeto.

A função abaixo recebe uma mensagem e um deslocamento e realiza essa substituição, mantendo letras maiúsculas e minúsculas.

```python
def criptografar_cesar(mensagem, deslocamento):
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():
            inicio = ord('A') if caractere.isupper() else ord('a')
            novo_caractere = chr((ord(caractere) - inicio + deslocamento) % 26 + inicio)
            resultado += novo_caractere
        else:
            resultado += caractere

    return resultado


# Solicita os dados ao usuário.
mensagem = input("Digite a mensagem: ")

# Adiciona tratamento de erro para a entrada do deslocamento
while True:
    try:
        deslocamento = int(input("Digite o deslocamento: "))
        break # Sai do loop se a conversão for bem-sucedida
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro para o deslocamento.")

# Criptografa a mensagem.
mensagem_criptografada = criptografar_cesar(mensagem, deslocamento)

# Exibe o resultado.
print("Mensagem criptografada:", mensagem_criptografada)
```

# Exercício 24 — Validação de CPF

### Enunciado

Implemente uma função para validar um CPF brasileiro.

### Explicação

Primeiro, removo caracteres que não sejam números para permitir que o CPF seja informado com ou sem pontuação.

Depois, verifico a quantidade de dígitos e calculo os dois dígitos verificadores do CPF. No final, comparo os dígitos calculados com os informados.

```python
def validar_cpf(cpf):
    # Remove tudo que não for número.
    cpf = ''.join(caractere for caractere in cpf if caractere.isdigit())

    # O CPF precisa possuir 11 dígitos.
    if len(cpf) != 11:
        return False

    # CPFs com todos os dígitos iguais são inválidos.
    if cpf == cpf[0] * 11:
        return False

    # Calcula o primeiro dígito verificador.
    soma = 0
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)

    resto = soma % 11
    primeiro_digito = 0 if resto < 2 else 11 - resto

    if primeiro_digito != int(cpf[9]):
        return False

    # Calcula o segundo dígito verificador.
    soma = 0
    for i in range(10):
        soma += int(cpf[i]) * (11 - i)

    resto = soma % 11
    segundo_digito = 0 if resto < 2 else 11 - resto

    return segundo_digito == int(cpf[10])


# Solicita o CPF ao usuário.
cpf = input("Digite o CPF: ")

# Verifica se o CPF é válido.
if validar_cpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")
```

# Exercício 30 — Jogo da Forca

### Enunciado

Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.

### Explicação

O programa possui uma palavra secreta e mostra ao jogador as letras que já foram descobertas.

A cada tentativa, o usuário informa uma letra. Caso ela esteja na palavra, suas posições são reveladas. O jogador possui um número limitado de erros.

```python
# Palavra que o jogador deverá descobrir.
palavra = "python"

# Cria uma lista com "_" para representar as letras escondidas.
descobertas = ["_"] * len(palavra)

# Controla as tentativas restantes.
tentativas = 6

print("Jogo da Forca!")
print("A palavra possui", len(palavra), "letras.")

while tentativas > 0 and "_" in descobertas:
    print("\nPalavra:", " ".join(descobertas))
    print("Tentativas restantes:", tentativas)

    letra = input("Digite uma letra: ").lower()

    if len(letra) != 1 or not letra.isalpha():
        print("Digite apenas uma letra.")
        continue

    if letra in palavra:
        # Revela todas as posições onde a letra aparece.
        for i in range(len(palavra)):
            if palavra[i] == letra:
                descobertas[i] = letra

        print("Você acertou uma letra!")
    else:
        tentativas -= 1
        print("A letra não está na palavra.")

if "_" not in descobertas:
    print("\nParabéns! Você descobriu a palavra:", palavra)
else:
    print("\nFim de jogo! A palavra era:", palavra)
```

# Exercício 31 — Ordenação por Bolha

### Enunciado

Implemente o algoritmo de ordenação por bolha (Bubble Sort).

### Explicação

O Bubble Sort compara elementos vizinhos e troca suas posições quando estão fora de ordem.

Esse processo é repetido várias vezes até que todos os elementos estejam organizados em ordem crescente.

```python
# Lista que será ordenada.
numeros = [5, 3, 8, 4, 2, 1]

# Percorre a lista várias vezes.
for i in range(len(numeros)):

    # Compara os elementos vizinhos.
    for j in range(0, len(numeros) - i - 1):

        # Troca os elementos se estiverem fora de ordem.
        if numeros[j] > numeros[j + 1]:
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

# Exibe a lista ordenada.
print("Lista ordenada:", numeros)
```

# Exercício 32 — Jogo da Velha

### Enunciado

Crie um jogo da velha para dois jogadores.

### Explicação

O jogo utiliza uma matriz 3×3 para representar o tabuleiro.

Os jogadores alternam entre `X` e `O`, escolhendo uma posição livre. Depois de cada jogada, o programa verifica se algum jogador conseguiu formar uma linha, coluna ou diagonal.

```python
# Cria o tabuleiro vazio.
tabuleiro = [[" " for _ in range(3)] for _ in range(3)]

# Define o jogador inicial.
jogador = "X"


def mostrar_tabuleiro():
    print()
    for linha in tabuleiro:
        print(" | ".join(linha))
        print("-" * 9)


def verificar_vencedor():
    # Verifica as linhas.
    for linha in tabuleiro:
        if linha[0] != " " and linha[0] == linha[1] == linha[2]:
            return True

    # Verifica as colunas.
    for coluna in range(3):
        if tabuleiro[0][coluna] != " " and \
           tabuleiro[0][coluna] == tabuleiro[1][coluna] == tabuleiro[2][coluna]:
            return True

    # Verifica a diagonal principal.
    if tabuleiro[0][0] != " " and \
       tabuleiro[0][0] == tabuleiro[1][1] == tabuleiro[2][2]:
        return True

    # Verifica a diagonal secundária.
    if tabuleiro[0][2] != " " and \
       tabuleiro[0][2] == tabuleiro[1][1] == tabuleiro[2][0]:
        return True

    return False


# Realiza até 9 jogadas.
for jogada in range(9):
    mostrar_tabuleiro()

    print("Jogador", jogador)

    linha = int(input("Digite a linha (1 a 3): ")) - 1
    coluna = int(input("Digite a coluna (1 a 3): ")) - 1

    # Verifica se a posição é válida.
    if linha < 0 or linha > 2 or coluna < 0 or coluna > 2:
        print("Posição inválida.")
        continue

    if tabuleiro[linha][coluna] != " ":
        print("Essa posição já está ocupada.")
        continue

    # Registra a jogada.
    tabuleiro[linha][coluna] = jogador

    # Verifica se houve vencedor.
    if verificar_vencedor():
        mostrar_tabuleiro()
        print("Jogador", jogador, "venceu!")
        break

    # Alterna o jogador.
    if jogador == "X":
        jogador = "O"
    else:
        jogador = "X"

else:
    mostrar_tabuleiro()
    print("O jogo terminou empatado.")
```

# Exercício 37 — Funções Recursivas

### Enunciado

Implemente uma função recursiva para calcular o fatorial de um número.

### Explicação

Uma função recursiva é uma função que chama a si mesma.

Neste caso, o fatorial pode ser definido como o número multiplicado pelo fatorial do número anterior. O caso base ocorre quando o número chega a 0 ou 1.

```python
def fatorial(numero):
    # Caso base da função recursiva.
    if numero <= 1:
        return 1

    # Chama a própria função com um número menor.
    return numero * fatorial(numero - 1)


# Solicita um número ao usuário.
numero = int(input("Digite um número inteiro: "))

# Calcula e exibe o fatorial.
print("O fatorial de", numero, "é:", fatorial(numero))
```

# Exercício 46 — Calcular Média e Desvio Padrão

### Enunciado

Calcule a média e o desvio padrão de uma lista de números.

### Explicação

Primeiro, calculo a média dos valores da lista.

Depois, calculo o desvio padrão utilizando a diferença entre cada valor e a média. A partir dessas diferenças, calculo a variância e, por fim, a raiz quadrada da variância.

```python
# Importa o módulo math para utilizar a raiz quadrada.
import math

# Lista de números.
numeros = [10, 20, 30, 40, 50]

# Calcula a média.
media = sum(numeros) / len(numeros)

# Calcula a soma dos quadrados das diferenças em relação à média.
soma_diferencas = 0

for numero in numeros:
    soma_diferencas += (numero - media) ** 2

# Calcula a variância.
variancia = soma_diferencas / len(numeros)

# Calcula o desvio padrão.
desvio_padrao = math.sqrt(variancia)

# Exibe os resultados.
print("Média:", media)
print("Desvio padrão:", desvio_padrao)
```

# Exercício 48 — Ordenação por Inserção

### Enunciado

Implemente o algoritmo de ordenação por inserção (Insertion Sort).

### Explicação

O Insertion Sort organiza a lista considerando uma parte já ordenada e inserindo cada novo elemento na posição correta.

A cada repetição, comparo o elemento atual com os anteriores e desloco os maiores para abrir espaço para ele.

```python
# Lista que será ordenada.
numeros = [8, 4, 3, 7, 6, 5, 2]

# Percorre a lista a partir do segundo elemento.
for i in range(1, len(numeros)):
    chave = numeros[i]
    j = i - 1

    # Move os elementos maiores uma posição para a direita.
    while j >= 0 and numeros[j] > chave:
        numeros[j + 1] = numeros[j]
        j -= 1

    # Coloca o elemento na posição correta.
    numeros[j + 1] = chave

# Exibe a lista ordenada.
print("Lista ordenada:", numeros)
```
