## Exercício 1: Integral de função trigonométrica
Calcule a integral da função f(x) = x²·sin(x) no intervalo de 0 a π.


```python
def f1(x): return x**2 * np.sin(x)
res1, err1 = integrate.quad(f1, 0, np.pi)
print("1 - Integral:", res1)
```

## Exercício 2: Equação diferencial
Resolva a equação diferencial dy/dx = x - y, com y(0) = 1, no intervalo de 0 a 10.


```python
def edo2(x, y): return x - y
sol2 = integrate.solve_ivp(edo2, [0, 10], [1], t_eval=np.linspace(0, 10, 100))
print("2 - Solução em t=10:", sol2.y[0][-1])
```

## Exercício 3: Raízes de função polinomial
Encontre as raízes da função f(x) = x⁴ - 3x³ + 2 utilizando SciPy.


```python
def f3(x): return x**4 - 3*x**3 + 2
raizes3 = optimize.root(f3, [0, 1, 2, 3])
print("3 - Raízes:", raizes3.x)
```

## Exercício 4: Maximização de função
Use a função de otimização de SciPy para maximizar f(x) = -x² + 4x + 1.


```python
def f4(x): return -( -x**2 + 4*x + 1 )
res4 = optimize.minimize(f4, 0)
print("4 - Máximo em x =", res4.x[0], "valor =", -res4.fun)
```

## Exercício 5: Transformada de Fourier
Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).


```python
t5 = np.linspace(0, 1, 500)
sinal5 = np.sin(2*np.pi*t5) + np.cos(4*np.pi*t5)
fft5 = fft.fft(sinal5)
print("5 - FFT (primeiros 5):", fft5[:5])
```

## Exercício 6: Matriz inversa

## Exercício 7: Interpolação cúbica
Use a função de interpolação cúbica de SciPy para interpolar os dados (0,0), (1,2), (2,3), (3,5), (4,4).


```python
x7 = np.array([0,1,2,3,4])
y7 = np.array([0,2,3,5,4])
f7 = interpolate.interp1d(x7, y7, kind='cubic')
x_novo7 = np.linspace(0, 4, 10)
y_novo7 = f7(x_novo7)
print("7 - Interpolados:", y_novo7)
```

## Exercício 8: Sistema de equações

```python
A8 = np.array([[3,2],[1,4]])
b8 = np.array([5,6])
sol8 = linalg.solve(A8, b8)
print("8 - Solução [x,y]:", sol8)
```

## Exercício 9: Ajuste de curva
Ajuste uma curva quadrática aos pontos (1,2), (2,4), (3,5), (4,7), (5,8).


```python
x9 = np.array([1,2,3,4,5])
y9 = np.array([2,4,5,7,8])
def quad(x, a, b, c): return a*x**2 + b*x + c
params9, _ = optimize.curve_fit(quad, x9, y9)
print("9 - Parâmetros [a,b,c]:", params9)
```

## Exercício 10: Integral dupla
Calcule a integral dupla da função f(x,y) = xy + x² no intervalo [0,2] para x e [0,1] para y.

```python
def f10(y, x): return x*y + x**2
res10, err10 = integrate.dblquad(f10, 0, 2, lambda x: 0, lambda x: 1)
print("10 - Integral dupla:", res10)
```

## Exercício 10: Três séries
Plote três gráficos de linha em um único gráfico, cada um com uma cor e estilo de linha diferentes.


```python
x10 = np.linspace(0, 10, 100)
y10a = np.sin(x10)
y10b = np.cos(x10)
y10c = np.sin(x10) + np.cos(x10)
plt.figure()
plt.plot(x10, y10a, 'r-', label='sin')
plt.plot(x10, y10b, 'g--', label='cos')
plt.plot(x10, y10c, 'b:', label='sin+cos')
plt.legend()
plt.title('Exercício 10 - Três séries')
plt.show()
```
